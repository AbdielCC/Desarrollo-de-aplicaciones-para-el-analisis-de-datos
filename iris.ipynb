```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Load the iris dataset
iris = load_iris()

# Convert to DataFrame
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df['target'] = iris.target
```

```python
iris_df
```

## Holdout estratificado

```python
X_train, X_test, y_train, y_test = train_test_split(iris_df, iris_df.target, random_state=42, test_size=0.3, stratify=iris_df.target)
```

```python
plt.scatter(X_train.iloc[:, 0], X_train.iloc[:, 1], c=y_train, cmap='viridis')
plt.xlabel('Característica 1')
plt.ylabel('Característica 2')
plt.title('Visualización de X_train')
plt.colorbar(label='Clase')
plt.show()
```
